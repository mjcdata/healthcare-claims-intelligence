## Step 2: Profile Members Completeness & Distributions

```python
# Profile missing values in Members dataset

missing_count = members_raw.isna().sum()
missing_pct = (members_raw.isna().mean() * 100).round(2)

members_missing = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percent": missing_pct
})

display(members_missing)
```

```python
# Profile categorical distributions in Members dataset

categorical_columns = [
    "gender",
    "state",
    "plan_type",
    "enrollment_status"
]

for column in categorical_columns:
    print(f"\n{column.upper()}")
    print(members_raw[column].value_counts(dropna=False))
```

```python
# Compare enrollment status with termination date completeness

pd.crosstab(
    members_raw["enrollment_status"],
    members_raw["termination_date"].isna(),
    rownames=["Enrollment Status"],
    colnames=["Termination Date Missing"]
)
```

```python
# Convert date fields for profiling only

date_columns = [
    "date_of_birth",
    "enrollment_date",
    "termination_date"
]

for column in date_columns:
    dates = pd.to_datetime(members_raw[column], errors="coerce")

    print(f"\n{column.upper()}")
    print("Earliest:", dates.min())
    print("Latest:", dates.max())
    print("Invalid/Unparseable:", dates.isna().sum() - members_raw[column].isna().sum())
```

```python
# Create temporary date fields for profiling

dob = pd.to_datetime(members_raw["date_of_birth"])
enrollment = pd.to_datetime(members_raw["enrollment_date"])
termination = pd.to_datetime(members_raw["termination_date"])

# Test logical date relationships

enrollment_before_birth = (enrollment < dob).sum()

termination_before_enrollment = (
    termination.notna() &
    (termination < enrollment)
).sum()

print("Enrollment Before Date of Birth:", enrollment_before_birth)
print("Termination Before Enrollment:", termination_before_enrollment)
```

```python
# Inspect members enrolled before their date of birth

invalid_enrollment_dates = members_raw[
    pd.to_datetime(members_raw["enrollment_date"]) <
    pd.to_datetime(members_raw["date_of_birth"])
]

print("Records with enrollment before birth:", len(invalid_enrollment_dates))
display(invalid_enrollment_dates.head(10))
```

```python
# Calculate member age as of the end of the dataset period

reference_date = pd.Timestamp("2025-12-31")
dob = pd.to_datetime(members_raw["date_of_birth"])

member_age = (
    reference_date.year
    - dob.dt.year
    - (
        (reference_date.month < dob.dt.month)
        | (
            (reference_date.month == dob.dt.month)
            & (reference_date.day < dob.dt.day)
        )
    )
)

print(member_age.describe())
```

```python
# Profile member age groups

age_groups = pd.cut(
    member_age,
    bins=[-1, 17, 34, 49, 64, 120],
    labels=["0-17", "18-34", "35-49", "50-64", "65+"]
)

print(age_groups.value_counts().sort_index())
```

## Step 4: Profile Claims Structure

```python
# Profile basic structure of Claims dataset

print("Shape:")
print(claims_raw.shape)

print("\nColumns:")
print(claims_raw.columns.tolist())

print("\nData Types:")
print(claims_raw.dtypes)

print("\nSample Records:")
display(claims_raw.head())
```

```python
# Validate Claims grain and composite key

print("Total Rows:")
print(len(claims_raw))

print("\nUnique Claim IDs:")
print(claims_raw["claim_id"].nunique())

print("\nMissing Claim IDs:")
print(claims_raw["claim_id"].isna().sum())

print("\nMissing Service Line Numbers:")
print(claims_raw["service_line_number"].isna().sum())

print("\nDuplicate Claim ID + Service Line:")
print(
    claims_raw.duplicated(
        subset=["claim_id", "service_line_number"]
    ).sum()
)
```

```python
# Profile missing values in Claims dataset

missing_count = claims_raw.isna().sum()
missing_pct = (claims_raw.isna().mean() * 100).round(2)

claims_missing = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percent": missing_pct
})

display(claims_missing)
```

```python
# Validate Claims referential integrity

invalid_members = ~claims_raw["member_id"].isin(members_raw["member_id"])
invalid_providers = ~claims_raw["provider_id"].isin(providers_raw["provider_id"])

print("Claims with Member IDs not found in Members:")
print(invalid_members.sum())

print("\nClaims with Provider IDs not found in Providers:")
print(invalid_providers.sum())
```

```python
# Profile Claims service dates

service_dates = pd.to_datetime(
    claims_raw["service_date"],
    errors="coerce"
)

print("Earliest Service Date:")
print(service_dates.min())

print("\nLatest Service Date:")
print(service_dates.max())

print("\nInvalid/Unparseable Service Dates:")
print(
    service_dates.isna().sum()
    - claims_raw["service_date"].isna().sum()
)
```

```python
# Validate claim dates against member enrollment periods

claims_enrollment_check = claims_raw.merge(
    members_raw[
        ["member_id", "enrollment_date", "termination_date"]
    ],
    on="member_id",
    how="left"
)

service_date = pd.to_datetime(
    claims_enrollment_check["service_date"]
)

enrollment_date = pd.to_datetime(
    claims_enrollment_check["enrollment_date"]
)

termination_date = pd.to_datetime(
    claims_enrollment_check["termination_date"]
)

claims_before_enrollment = (
    service_date < enrollment_date
).sum()

claims_after_termination = (
    termination_date.notna()
    & (service_date > termination_date)
).sum()

print("Claims Before Enrollment:")
print(claims_before_enrollment)

print("\nClaims After Termination:")
print(claims_after_termination)
```

```python
# Measure overall enrollment-period violations

before_enrollment = service_date < enrollment_date

after_termination = (
    termination_date.notna()
    & (service_date > termination_date)
)

any_enrollment_violation = (
    before_enrollment | after_termination
)

print("Claims Before Enrollment:")
print(before_enrollment.sum())

print("\nClaims After Termination:")
print(after_termination.sum())

print("\nClaims Violating Either Rule:")
print(any_enrollment_violation.sum())

print("\nPercent of Claims Violating Either Rule:")
print(
    round(any_enrollment_violation.mean() * 100, 2)
)
```

```python
# Review examples of enrollment-period violations

print("CLAIMS BEFORE ENROLLMENT")
display(
    claims_enrollment_check.loc[
        before_enrollment,
        [
            "claim_id",
            "member_id",
            "service_date",
            "enrollment_date",
            "termination_date"
        ]
    ].head(10)
)

print("\nCLAIMS AFTER TERMINATION")
display(
    claims_enrollment_check.loc[
        after_termination,
        [
            "claim_id",
            "member_id",
            "service_date",
            "enrollment_date",
            "termination_date"
        ]
    ].head(10)
)
```

```python
# Profile allowed amount

print("Allowed Amount Summary:")
print(claims_raw["allowed_amount"].describe())

print("\nNegative Allowed Amounts:")
print((claims_raw["allowed_amount"] < 0).sum())

print("\nZero Allowed Amounts:")
print((claims_raw["allowed_amount"] == 0).sum())

print("\nPositive Allowed Amounts:")
print((claims_raw["allowed_amount"] > 0).sum())
```

```python
# Inspect highest allowed amounts

display(
    claims_raw[
        [
            "claim_id",
            "procedure_code",
            "procedure_description",
            "service_category",
            "place_of_service",
            "allowed_amount"
        ]
    ]
    .sort_values(
        "allowed_amount",
        ascending=False
    )
    .head(10)
)
```

```python
# Compare average allowed amount ratio by network status

network_summary = (
    network_check
    .groupby("network_status")
    .agg(
        claim_lines=("claim_id", "count"),
        avg_baseline=("baseline_allowed_amount", "mean"),
        avg_allowed=("allowed_amount", "mean"),
        avg_ratio=("baseline_ratio", "mean")
    )
)

display(network_summary)
```

## Step 6: Utilization, Cost, & Outlier Profiling

```python
# Profile claim utilization by member

member_utilization = (
    claims_raw
    .groupby("member_id")
    .agg(
        claim_lines=("claim_id", "size"),
        unique_claims=("claim_id", "nunique"),
        total_allowed=("allowed_amount", "sum"),
        avg_allowed=("allowed_amount", "mean")
    )
)

print("Members with Claims:")
print(len(member_utilization))

print("\nMember Utilization Summary:")
display(
    member_utilization[
        ["claim_lines", "unique_claims", "total_allowed", "avg_allowed"]
    ].describe()
)
```

```python
# Examine upper-end member utilization and cost

percentiles = [0.90, 0.95, 0.99]

print("Member Utilization Upper Percentiles:")
display(
    member_utilization[
        ["claim_lines", "unique_claims", "total_allowed", "avg_allowed"]
    ].quantile(percentiles)
)
```

```python
# Identify members in the top 1% by total allowed amount

cost_threshold_99 = member_utilization["total_allowed"].quantile(0.99)

high_cost_members = (
    member_utilization[
        member_utilization["total_allowed"] >= cost_threshold_99
    ]
    .sort_values("total_allowed", ascending=False)
)

print("99th Percentile Total Allowed Threshold:")
print(round(cost_threshold_99, 2))

print("\nNumber of High-Cost Members:")
print(len(high_cost_members))

print("\nTop 20 High-Cost Members:")
display(high_cost_members.head(20))
```

```python
# Analyze service categories for top 1% high-cost members

high_cost_claims = claims_raw[
    claims_raw["member_id"].isin(high_cost_members.index)
]

high_cost_service_summary = (
    high_cost_claims
    .groupby("service_category")
    .agg(
        claim_lines=("claim_id", "size"),
        total_allowed=("allowed_amount", "sum"),
        avg_allowed=("allowed_amount", "mean")
    )
    .sort_values("total_allowed", ascending=False)
)

high_cost_service_summary["percent_of_cost"] = (
    high_cost_service_summary["total_allowed"]
    / high_cost_service_summary["total_allowed"].sum()
    * 100
)

display(high_cost_service_summary)
```

```python
# Profile claim utilization and cost by provider

provider_utilization = (
    claims_raw
    .groupby("provider_id")
    .agg(
        claim_lines=("claim_id", "size"),
        unique_claims=("claim_id", "nunique"),
        unique_members=("member_id", "nunique"),
        total_allowed=("allowed_amount", "sum"),
        avg_allowed=("allowed_amount", "mean")
    )
)

print("Providers with Claims:")
print(len(provider_utilization))

print("\nProvider Utilization Summary:")
display(provider_utilization.describe())
```

```python
# Analyze monthly claim volume and allowed cost

monthly_trend = claims_raw.copy()

monthly_trend["service_date"] = pd.to_datetime(
    monthly_trend["service_date"]
)

monthly_trend["service_month"] = (
    monthly_trend["service_date"]
    .dt.to_period("M")
)

monthly_summary = (
    monthly_trend
    .groupby("service_month")
    .agg(
        claim_lines=("claim_id", "size"),
        unique_members=("member_id", "nunique"),
        total_allowed=("allowed_amount", "sum"),
        avg_allowed=("allowed_amount", "mean")
    )
    .reset_index()
)

display(monthly_summary)
```

```python
# Compare monthly claim volume with enrolled member population

members_monthly = members_raw.copy()

members_monthly["enrollment_date"] = pd.to_datetime(
    members_monthly["enrollment_date"]
)

members_monthly["termination_date"] = pd.to_datetime(
    members_monthly["termination_date"],
    errors="coerce"
)

months = pd.period_range("2021-01", "2025-12", freq="M")

monthly_enrollment = []

for month in months:
    month_end = month.to_timestamp(how="end")

    enrolled = (
        (members_monthly["enrollment_date"] <= month_end)
        &
        (
            members_monthly["termination_date"].isna()
            | (members_monthly["termination_date"] >= month_end)
        )
    )

    monthly_enrollment.append({
        "service_month": month,
        "enrolled_members": enrolled.sum()
    })

monthly_enrollment = pd.DataFrame(monthly_enrollment)

monthly_population_check = monthly_summary.merge(
    monthly_enrollment,
    on="service_month",
    how="left"
)

monthly_population_check["claim_lines_per_1000_members"] = (
    monthly_population_check["claim_lines"]
    / monthly_population_check["enrolled_members"]
    * 1000
)

display(
    monthly_population_check[
        [
            "service_month",
            "claim_lines",
            "enrolled_members",
            "claim_lines_per_1000_members"
        ]
    ]
)
```

```python
# Compare claim volume by calendar year

yearly_utilization = claims_raw.copy()

yearly_utilization["service_date"] = pd.to_datetime(
    yearly_utilization["service_date"]
)

yearly_utilization["service_year"] = (
    yearly_utilization["service_date"].dt.year
)

yearly_summary = (
    yearly_utilization
    .groupby("service_year")
    .agg(
        claim_lines=("claim_id", "size"),
        unique_claims=("claim_id", "nunique"),
        unique_members=("member_id", "nunique"),
        total_allowed=("allowed_amount", "sum")
    )
)

yearly_summary["claim_lines_per_member"] = (
    yearly_summary["claim_lines"]
    / yearly_summary["unique_members"]
)

display(yearly_summary)
```

```python
# Analyze service category mix by year

claims_time_mix = claims_raw.copy()

claims_time_mix["service_date"] = pd.to_datetime(
    claims_time_mix["service_date"]
)

claims_time_mix["service_year"] = (
    claims_time_mix["service_date"].dt.year
)

year_category_mix = (
    claims_time_mix
    .groupby(["service_year", "service_category"])
    .size()
    .reset_index(name="claim_lines")
)

year_category_mix["percent_of_year"] = (
    year_category_mix["claim_lines"]
    / year_category_mix.groupby("service_year")["claim_lines"].transform("sum")
    * 100
)

year_category_pivot = year_category_mix.pivot(
    index="service_category",
    columns="service_year",
    values="percent_of_year"
)

display(year_category_pivot.round(2))
```

```python
# Analyze service category cost mix by year

year_category_cost = (
    claims_time_mix
    .groupby(["service_year", "service_category"])
    ["allowed_amount"]
    .sum()
    .reset_index(name="total_allowed")
)

year_category_cost["percent_of_year_cost"] = (
    year_category_cost["total_allowed"]
    / year_category_cost.groupby("service_year")["total_allowed"].transform("sum")
    * 100
)

year_category_cost_pivot = year_category_cost.pivot(
    index="service_category",
    columns="service_year",
    values="percent_of_year_cost"
)

display(year_category_cost_pivot.round(2))
```
